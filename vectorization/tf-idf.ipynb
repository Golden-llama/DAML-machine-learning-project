```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
df = pd.read_csv('data/lemmatized.csv')
df.head()
```

```python
vectorizer = TfidfVectorizer(
    max_features=2000,    
    ngram_range=(1, 2),
    stop_words='english'
)
```

```python
df['text'] = df['text'].fillna('')
tfidf_matrix = vectorizer.fit_transform(df['text'])

tfidf_df = pd.DataFrame(
    tfidf_matrix.toarray(),
    columns=vectorizer.get_feature_names_out()
)
df_tfidf = pd.concat([df.reset_index(drop=True), tfidf_df.reset_index(drop=True)], axis=1)

df_tfidf = df_tfidf.drop(columns=['Unnamed: 0', 'title', 'text', 'label'])
```

```python
df_tfidf.to_csv('data/tfidf_dataset.csv', index=False)
df_tfidf.head()
```
